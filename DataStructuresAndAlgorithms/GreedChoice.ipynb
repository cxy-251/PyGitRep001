| 问题名称                | 贪心策略简述                |
| ------------------- | --------------------- |
| 零钱兑换（无限币）           | 每次找最大面额的钱             |
| 活动选择 / 区间调度         | 按结束时间排序，每次选不冲突的下一个    |
| 分数背包                | 每次选单位价值最高的物品（可分割）     |
| Huffman 编码          | 每次合并最小的两个频率节点         |
| 最小生成树（Prim/Kruskal） | 每次选权值最小的边，避免成环        |
| Dijkstra 最短路        | 每次选当前最短路径的点进行扩展       |
| 跳跃游戏 Jump Game      | 每次跳得最远，保证能跳到终点        |
| 加油站/区间覆盖类问题         | 每次在能到达的范围内选加油最多/覆盖最广的 |
