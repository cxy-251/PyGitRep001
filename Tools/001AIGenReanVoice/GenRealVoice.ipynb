````python
# edge.tts AI味儿重
import nest_asyncio
nest_asyncio.apply()

import asyncio
import edge_tts

async def tts(text, file_name="output.mp3"):
    communicate = edge_tts.Communicate(
        text=text,
        voice="zh-CN-XiaoxiaoNeural",
        rate="+0%"
    )
    await communicate.save(file_name)
    print(f"✅ 音频保存为 {file_name}")

# 调用
text = "臣亮言：先帝创业未半而中道崩殂，今天下三分，益州疲弊。"
asyncio.run(tts(text, "zhugeliang.mp3"))
````

````python
import os
import torch
from ChatTTS import Chat

# 打印路径，确保路径正确
gpt_ckpt_path = r"C:\02Programmer\02Proj\PyVSCode\GitPackage\ChatTTS\asset\gpt\model.safetensors"
dvae_ckpt_path = r"C:\02Programmer\02Proj\PyVSCode\GitPackage\ChatTTS\asset\dvae.safetensors"
vocos_ckpt_path = r"C:\02Programmer\02Proj\PyVSCode\GitPackage\ChatTTS\asset\vocos.safetensors"

# 确保文件存在
def check_file_exists(file_path):
    if os.path.exists(file_path):
        print(f"文件存在: {file_path}")
    else:
        print(f"文件不存在: {file_path}")
        raise FileNotFoundError(f"{file_path} 文件不存在!")

check_file_exists(gpt_ckpt_path)
check_file_exists(dvae_ckpt_path)
check_file_exists(vocos_ckpt_path)

# 初始化 ChatTTS
chat = Chat()

# 尝试加载模型
try:
    chat._load(
        gpt_ckpt_path=gpt_ckpt_path,
        dvae_ckpt_path=dvae_ckpt_path,
        vocos_ckpt_path=vocos_ckpt_path
    )
    print("模型加载成功")
except Exception as e:
    print(f"加载模型时出错: {e}")

# 可选：设置随机种子，确保每次生成一致
chat.set_random_seed(42)

# 输入文本（生成语音）
texts = ["今天天气真好呀，你打算去哪儿玩呢？"]

# 语音合成（use_decoder=True 会生成更自然的声音，但可能耗时更久）
wav_chunks = chat.infer(texts, use_decoder=True)

# 检查 wav_chunks 是否有效
if wav_chunks is not None and len(wav_chunks) > 0:
    # 保存为 WAV 文件
    import scipy.io.wavfile
    scipy.io.wavfile.write("output.wav", rate=24000, data=wav_chunks[0])
    print("✅ 已保存到 output.wav")
else:
    print("❌ 生成的音频无效，请检查模型加载和语音合成过程。")
````
