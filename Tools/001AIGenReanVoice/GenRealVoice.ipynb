````python
# edge.tts AI味儿重
import nest_asyncio
nest_asyncio.apply()

import asyncio
import edge_tts

async def tts(text, file_name="output.mp3"):
    communicate = edge_tts.Communicate(
        text=text,
        voice="zh-CN-XiaoxiaoNeural",
        rate="+0%"
    )
    await communicate.save(file_name)
    print(f"✅ 音频保存为 {file_name}")

# 调用
text = "臣亮言：先帝创业未半而中道崩殂，今天下三分，益州疲弊。"
asyncio.run(tts(text, "zhugeliang.mp3"))
````

````python
# import sys
# sys.path.append("C:\\02Programmer\\02Proj\\PyVSCode\\GitPackage\\ChatTTS")

import ChatTTS
import torch
import scipy.io.wavfile

chat = ChatTTS.Chat()
chat.load_models()

# 文本输入（可中文）
texts = ["今天天气真好呀，你打算去哪儿玩呢？"]

# 合成语音
wav = chat.infer(texts, use_decoder=True)

# 保存音频
scipy.io.wavfile.write("output.wav", rate=24000, data=wav[0])
````
