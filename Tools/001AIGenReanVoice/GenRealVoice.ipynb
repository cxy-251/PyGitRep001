````python
# edge.tts AI味儿重
import nest_asyncio
nest_asyncio.apply()

import asyncio
import edge_tts

async def tts(text, file_name="output.mp3"):
    communicate = edge_tts.Communicate(
        text=text,
        voice="zh-CN-XiaoxiaoNeural",
        rate="+0%"
    )
    await communicate.save(file_name)
    print(f"✅ 音频保存为 {file_name}")

# 调用
text = "臣亮言：先帝创业未半而中道崩殂，今天下三分，益州疲弊。"
asyncio.run(tts(text, "zhugeliang.mp3"))
````

````python
import torch
import scipy.io.wavfile
from ChatTTS import Chat

# 初始化 ChatTTS 对象
chat = Chat()

# 正确加载模型
chat._load(vocos_ckpt_path='C:\\02Programmer\\02Proj\\PyVSCode\\GitPackage\\ChatTTS\\asset\\Vocos.pt',)

# 可选：设定随机种子（保证每次生成一致）
chat.set_random_seed(42)

# 输入文本（支持中文）
texts = ["今天天气真好呀，你打算去哪儿玩呢？"]

# 语音合成（use_decoder=True 会生成更自然的声音，但可能耗时更久）
wav_chunks = chat.infer(texts, use_decoder=True)

# 保存为 WAV 文件
scipy.io.wavfile.write("output.wav", rate=24000, data=wav_chunks[0])
print("✅ 已保存到 output.wav")
````
